# Development of the Bayesian model to characterize the reliability of SC on synthetic data

Thinking about the problem, I converged to the following model of the density distribution:

p(D) = p(D| c=0)p(c=0) + p(D| c=1)p(c=1)

     = pi_0 * Exp(lambda) + (1-pi_0) * TruncatedNormal(mu, sigma)

with 

pi_0 ~ Beta(1,1)

lambda ~ Gamma(2,3)

sigma ~ HalfNormal(0,0.1)

```python
import pickle
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from bayesian_modeling import run_simulation

output_dir = "/home/cprovins/projects/bayesian_sc/mixture_model_lambdasigma"
```

```python
results_fixed = run_simulation(
    true_pi0=0.9,
    true_lambda=2.0,
    true_mu=0.8,
    true_sigma=0.15,
    n_samples=36,
    draws=2000,
    display_plots=True,
    random_seed=None,
    repeat_fit=30,
)
# Save the results
params = results_fixed["data"]["params"]
filename = f"results_fixed_pi0_{params['pi0']}_lambda_{params['lambda_exp']}_mu_{params['mu']}_sigma_{params['sigma']}.pkl"
with open(os.path.join(output_dir, filename), "wb") as f:
    pickle.dump(results_fixed, f)
```

```python
def plot_summary_stats(results_multi, param_values, param_varied):
    param_names = ["pi0", "sigma"]
    param_names_str = ["$\pi_0$", "$\sigma$"]
    fig, axes = plt.subplots(1, len(param_names), figsize=(20, 6))
    for i, param_name in enumerate(param_names):
        # Extract RMSE values
        rmse_values = [
            results_multi[p]["summary_stats"].loc[param_name, "rmse"]
            for p in param_values
        ]

        # Plot RMSE against param_varied
        ax = axes[i]
        ax.plot(param_values, rmse_values, marker="o", linestyle="-", color="b")
        if param_varied == "$\mu$":
            ax.set_xscale("log")
        ax.set_xticks(param_values)
        ax.tick_params(axis="x", labelsize=12)
        ax.tick_params(axis="y", labelsize=12)
        ax.set_xlabel(param_varied, fontsize=16)
        ax.set_ylabel(f"RMSE on estimated {param_names_str[i]}", fontsize=14)
        ax.grid(True)
        ax.set_title(f"RMSE for {param_names_str[i]}", fontsize=16)

    true_values = [
        [results_multi[p]["data"]["params"][param_name] for p in param_values][0]
        for param_name in param_names
    ]
    for param_name, true_value in zip(param_names_str, true_values):
        # Add a textbox to the figure
        textstr = "Parameter\ntrue values:\n" + "\n".join(
            [
                f"{param_name}: {true_value}"
                for param_name, true_value in zip(param_names_str, true_values)
            ]
        )
        props = dict(boxstyle="round", facecolor="wheat", alpha=0.5)
        axes[-1].text(
            1.05,
            0.5,
            textstr,
            transform=axes[-1].transAxes,
            fontsize=14,
            verticalalignment="center",
            bbox=props,
        )
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(1, len(param_names), figsize=(20, 6))
    for i, param_name in enumerate(param_names):
        # Extract RMSE values
        rmse_values = [
            results_multi[p]["summary_stats"].loc[param_name, "rmse"]
            for p in param_values
        ]
        relative_rmse_values = np.array(rmse_values) / np.array(param_values) * 100

        # Plot RMSE against param_varied
        ax = axes[i]
        ax.plot(
            param_values, relative_rmse_values, marker="o", linestyle="-", color="b"
        )
        if param_varied == "$\mu$":
            ax.set_xscale("log")
        ax.set_xticks(param_values)
        ax.tick_params(axis="x", labelsize=12)
        ax.tick_params(axis="y", labelsize=12)
        ax.set_xlabel(param_varied, fontsize=16)
        ax.set_ylabel(f"Relative RMSE on estimated {param_names_str[i]} %", fontsize=14)
        ax.grid(True)
        ax.set_title(f"Relative RMSE for {param_names_str[i]}", fontsize=16)

    true_values = [
        [results_multi[p]["data"]["params"][param_name] for p in param_values][0]
        for param_name in param_names
    ]
    for param_name, true_value in zip(param_names_str, true_values):
        # Add a textbox to the figure
        textstr = "Parameter\ntrue values:\n" + "\n".join(
            [
                f"{param_name}: {true_value}"
                for param_name, true_value in zip(param_names_str, true_values)
            ]
        )
        props = dict(boxstyle="round", facecolor="wheat", alpha=0.5)
        axes[-1].text(
            1.05,
            0.5,
            textstr,
            transform=axes[-1].transAxes,
            fontsize=14,
            verticalalignment="center",
            bbox=props,
        )
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(1, len(param_names), figsize=(20, 6))
    for i, param_name in enumerate(param_names):
        # Extract mean parameter estimation
        means_param = [
            results_multi[p]["all_summaries"][
                results_multi[p]["all_summaries"]["param"] == param_name
            ]["mean"].mean()
            for p in param_values
        ]

        # Plot mean param_name against param_varied
        ax = axes[i]
        ax.plot(param_values, means_param, marker="o", linestyle="-", color="b")
        if param_varied == "$\mu$":
            ax.set_xscale("log")
        ax.set_xticks(param_values)
        ax.tick_params(axis="x", labelsize=12)
        ax.tick_params(axis="y", labelsize=12)
        ax.set_xlabel(param_varied, fontsize=16)
        ax.set_ylabel(f"Mean estimated {param_names_str[i]}", fontsize=14)
        ax.grid(True)
        ax.set_title(f"Mean {param_names_str[i]}", fontsize=16)

    true_values = [
        [results_multi[p]["data"]["params"][param_name] for p in param_values][0]
        for param_name in param_names
    ]
    for param_name, true_value in zip(param_names_str, true_values):
        # Add a textbox to the figure
        textstr = "Parameter\ntrue values:\n" + "\n".join(
            [
                f"{param_name}: {true_value}"
                for param_name, true_value in zip(param_names_str, true_values)
            ]
        )
        props = dict(boxstyle="round", facecolor="wheat", alpha=0.5)
        axes[-1].text(
            1.05,
            0.5,
            textstr,
            transform=axes[-1].transAxes,
            fontsize=14,
            verticalalignment="center",
            bbox=props,
        )
    plt.tight_layout()
    plt.show()

    # # Extract estimation error for each pi0
    # for p in results_multi.keys():
    #     # Aggregate the dataframes for each p into a single dataframe
    #     if p == param_values[0]:
    #         df_agg = results_multi[p]["all_summaries"][results_multi[p]["all_summaries"]["param"] == param_name].copy()
    #         df_agg[param_name] = p
    #     else:
    #         df_tmp = results_multi[p]["all_summaries"][results_multi[p]["all_summaries"]["param"] == param_name].copy()
    #         df_tmp[param_name] = p
    #         df_agg = pd.concat([df_agg, df_tmp], ignore_index=True)
    # df = df_agg
    # # Plot scatter plot of pi0 versus estimation error
    # plt.figure(figsize=(8, 6))
    # plt.scatter(df["true_value"], df["estimation_error"], color='r', label='Estimation Error')
    # plt.xlabel(param_name)
    # plt.ylabel("Estimation Error")
    # plt.title(f"Scatter Plot of {param_name} vs Estimation Error")
    # plt.grid(True)
    # plt.legend()
    # plt.show()
```
